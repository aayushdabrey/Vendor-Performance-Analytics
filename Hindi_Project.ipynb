```python
import pandas as pd 
import os 
from sqlalchemy import create_engine
```

```python
engine = create_engine('sqlite:///inventory.db')
```

```python
for file in os.listdir('data'):
    if '.csv' in file:
        df = pd.read_csv('data/' + file)
        print(df.shape)
        ingest_db(df, file[:-4], engine)
```

```python
def ingest_db(df, table_name, engine):
    df.to_sql(
        table_name,
        con=engine,
        if_exists='replace',
        index=False
    )
```

```python
query = """
SELECT
VendorNumber,
VendorName,
SUM(Dollars) AS PurchaseAmount,
SUM(Quantity) AS PurchaseQty

FROM purchases

GROUP BY VendorNumber, VendorName
"""

vendor_summary = pd.read_sql(query, conn)

vendor_summary.head()
```

```python
top_vendors = vendor_summary.sort_values(
    by="PurchaseAmount",
    ascending=False
)

top_vendors.head(10)
```
